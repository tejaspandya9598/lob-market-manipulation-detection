# 01 — Data Preparation
Load raw CSVs, fill missing values, engineer features, cache as parquet for downstream notebooks.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent
sys.path.insert(0, str(ROOT))

import yaml
import pandas as pd
from src.features import build_features

cfg = yaml.safe_load((ROOT / 'configs' / 'config.yaml').read_text())
raw_cols = cfg['features']['raw']
raw_dir = ROOT / cfg['paths']['raw_dir']
```

```python
def load(name):
    df = pd.read_csv(raw_dir / name)
    df[raw_cols] = df[raw_cols].fillna(0.0)
    return build_features(df)

train = load(cfg['data']['train_file'])
val = load(cfg['data']['val_file'])
test = load(cfg['data']['test_file'])
train.shape, val.shape, test.shape
```

```python
out = ROOT / 'data' / 'processed'
out.mkdir(parents=True, exist_ok=True)
train.to_parquet(out / 'train.parquet', index=False)
val.to_parquet(out / 'val.parquet', index=False)
test.to_parquet(out / 'test.parquet', index=False)
```
